```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.MultiDiGraph()
G.add_nodes_from(["A1", "A2", "A3", "B", "C1", "C2", "C3"])
G.add_edges_from([("A1","B"),("A2","B"),("A3","B"),("B","C1"),("B","C2"), ("B","C3")])
nx.draw(G, with_labels=True)
plt.title("ABC Model")
plt.show()
# want clumps / organized - can I adjust the location of a node?
# how to get from just the top n from a json or sqlite file 
```

```python
H = nx.MultiDiGraph()
H.add_nodes_from(["A", "B1", "B2", "C"])
H.add_edges_from([("A","B1"),("B1","B2"),("B2","C") ])
nx.draw(H, with_labels=True)
plt.title("ANC Model")
plt.show()
```

```python
I = nx.MultiDiGraph()
I.add_nodes_from(["A", "B", "C"])
I.add_edges_from([("A","B"),("B","C"),("A","C") ])
nx.draw(I, with_labels=True)
plt.title("Direction Connection between two papers - would ignore. Eg same author")
plt.show()
```
